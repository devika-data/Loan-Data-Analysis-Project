```python
import pandas as pd

df = pd.read_csv('Cleaned-Loan-Data-New-Jupiter.csv')
print(df.head())
```

```python
df.isnull().sum()
```

```python
df['Loan_Amount_Term'] = df['Loan_Amount_Term'].astype(int)
df['Credit_History'] = df['Credit_History'].astype(int)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.to_csv('Final-cleaned-New-Loandata.csv' , index=False)
```

```python
# 🔷 Step 1: Understand Target Variable (Loan_Status)
df['Loan_Status'].value_counts()
df['Loan_Status'].value_counts(normalize=True) * 100
```

```python
df['Loan_Status'].value_counts(normalize=True).mul(100).round(1)
```

```python
#🔷 Step 2: Credit History (MOST IMPORTANT FEATURE)

import seaborn as sns
sns.countplot(x='Credit_History', hue='Loan_Status', data=df)
```

```python
# Credit history is the strongest factor influencing loan approval.
```

```python
 #🔷 Step 3: Income Analysis

df['TotalIncome'] = df['ApplicantIncome'] + df['CoapplicantIncome']

sns.histplot(df['TotalIncome'],kde=True)
plt.show()
```

```python
#Income distribution is positively skewed, indicating most applicants fall in lower income range.
```

```python
# 🔷 Step 4: Income vs Loan Status
sns.boxplot(x='Loan_Status', y='TotalIncome', data=df)
```

```python
# Higher income slightly increases approval chances.
```

```python
# 🔷 Step 5: Property Area Impact

sns.countplot(x='Property_Area', hue='Loan_Status', data=df)
plt.title("Property Area Impact on Loan Approval")
plt.xlabel("Property_Area")
plt.ylabel("Loan_Status_count")
plt.show()
```

```python
#Loan approvals vary by property area, with semi-urban regions showing higher approvals.
```

```python
# 🔷 Step 6: Education Impact

sns.countplot(x='Education', hue='Loan_Status', data=df)
plt.title("Educational Impact on Loan Approval")
plt.xlabel("Education")
plt.ylabel("Loan_Status_count")
plt.show()
```

```python
# Graduate applicants get more loan approvals than non-graduate applicants.
```

```python
# 🔷 Step 7: Dependents Impact

sns.countplot(x='Dependents', hue='Loan_Status', data=df)
plt.title("Dependents Impact")
plt.xlabel("Dependents")
plt.ylabel("Loan_Status")
plt.tight_layout()
plt.show()
```

```python
#Loan approval is higher for applicants with fewer dependents.
```

```python
# $ 🔷 Step 8: Loan Amount Analysis

sns.histplot(df['LoanAmount'], kde=True)
plt.show()
```

```python
# LoanAmount distribution is skewed, justifying median imputation.
```

```python
# 🔷 Step 8: Correlation Heatmap

sns.heatmap(df.corr(numeric_only=True), annot=True)

plt.title("Correlation Heatmap of Loan Features")
plt.show()
```

```python
# Credit History is an important factor for loan approval.
```

✅ 1. Credit History is Important
			Applicants with a good credit history are more likely to get their loans approved.


✅ 2. Income Supports Approval
 			Higher income helps in loan approval, but it is not the only important factor.

✅ 3. Property Area Affects Approval
 			Semi-urban areas show a better loan approval trend compared to other areas.

✅ 4. Rural Areas Have Growth Potential
		 Rural areas can be a good opportunity for business growth, but they involve higher risk. A careful lending strategy with proper credit checks can help manage this risk.
